Para el entrenamiento se realizo una division de imagene en la base de datos exportada desde roboflow en formato YoloV8 la cual quedo dividido en train/valid/test con images y labels en sus repectivas carpetas.

```python
import os
import random
import shutil

# Carpeta base del dataset YOLOv8
DATASET_DIR = "Yoni4Vision.yolov8/train"
IMG_DIR = os.path.join(DATASET_DIR, "images")
LBL_DIR = os.path.join(DATASET_DIR, "labels")

# Proporciones
TRAIN_RATIO = 0.7
VAL_RATIO = 0.2
TEST_RATIO = 0.1

# Listar imágenes
images = [f for f in os.listdir(IMG_DIR) if f.endswith((".jpg", ".png"))]
random.shuffle(images)

n = len(images)
train_imgs = images[:int(n*TRAIN_RATIO)]
val_imgs   = images[int(n*TRAIN_RATIO):int(n*(TRAIN_RATIO+VAL_RATIO))]
test_imgs  = images[int(n*(TRAIN_RATIO+VAL_RATIO)):]

splits = {"train": train_imgs, "valid": val_imgs, "test": test_imgs}

# Crear carpetas destino
OUTPUT_DIR = "Yoni4Vision.yolov8_split"
for split in splits.keys():
    os.makedirs(os.path.join(OUTPUT_DIR, split, "images"), exist_ok=True)
    os.makedirs(os.path.join(OUTPUT_DIR, split, "labels"), exist_ok=True)

# Copiar imágenes y labels
for split, imgs in splits.items():
    for img in imgs:
        # Imagen
        shutil.copy(os.path.join(IMG_DIR, img),
                    os.path.join(OUTPUT_DIR, split, "images", img))
        # Label correspondiente
        label_file = img.rsplit(".", 1)[0] + ".txt"
        src_lbl = os.path.join(LBL_DIR, label_file)
        if os.path.exists(src_lbl):
            shutil.copy(src_lbl,
                        os.path.join(OUTPUT_DIR, split, "labels", label_file))

print("✅ Dataset dividido en train/valid/test con images y labels")
```
